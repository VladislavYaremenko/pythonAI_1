```python
temperatures = [10, -3, 6, 2, 7, 3, 1]
```

```python
average = sum(temperatures)/len(temperatures)
maximum = max(temperatures)
minimum = min(temperatures)
difference = maximum - minimum

print("Середнє значення:", average)
print("Максимальне значення:", maximum)
print("Мінімальне значення:", minimum)
print("Різниця:", difference)
```

| Показник | Значення |
|---|---:|
| Середнє значення | 4.0 |
| Максимальне значення | 7 |
| Мінімальне значення | 1 |
| Різниця між максимальним і мінімальним | 6 |
