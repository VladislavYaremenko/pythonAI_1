```python
costs = [1200, 1500, 800, 600, 1000, 400, 900]
suma = 0
quantity = len(costs)
for el in costs:
    suma += el
print("Загальна сума:", suma)
average = suma / quantity
print("Середнє значення:", average)
maximum = max(costs)
minimum = min(costs)
print("Максимальна витрата:", maximum)
print("Мінімальна витрата:", minimum)

above1000 = 0

for el in costs:
    if el >= 1000:
        above1000 += 1

print("Кількість днів з витратами понад 1000:", above1000)
```
