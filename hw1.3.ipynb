```python
from collections import Counter
votes = ["apple", "banana", "apple", "orange", "banana", "apple", "apple", "apple", "orange"]
counts = Counter(votes)
print(counts)
leader = counts.most_common(1)[0]
print("Лідер:", leader[0])
print("Кількість голосів:", leader[1])
top_3 = counts.most_common(3)
print("Топ-3:")
for fruit, count in top_3:
    print(fruit, "->", count)
```

| Фрукт | Кількість голосів |
|---|---:|
| Apple | 5 |
| Banana | 2 |
| Orange | 2 |

### Лідер

**Apple — 5 голосів**
