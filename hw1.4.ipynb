```python
math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

math_average = sum(math) / len(math)
physics_average = sum(physics) / len(physics)
english_average = sum(english) / len(english)

math_min = min(math)
physics_min = min(physics)
english_min = min(english)

math_max = max(math)
physics_max = max(physics)
english_max = max(english)

averages = {
    "Math": math_average,
    "Physics": physics_average,
    "English": english_average
}
best_subject = max(averages, key=averages.get)

print("Math:", math_average, math_min, math_max)
print("Physics:", physics_average, physics_min, physics_max)
print("English:", english_average, english_min, english_max)

print("Предмет з найвищим середнім результатом:", best_subject)
```
